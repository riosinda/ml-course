### 2. Probabilidad binomial
### Una familia tiene tres hijos. Suponiendo que cada hijo tiene un 50% de probabilidad de ser varón, independientemente de los demás, responde:

Datos:
    $n=3$,
    $p=0.5$,
    $q=1-p=0.5$

**a) ¿Cuál es la probabilidad de tener exactamente dos hijos varones?**

Utilizando la formula $P(X=k)= \binom{n}{k}p^k(1-p)^{n-k}$ y sustituyendo los valores, tenemos que $P(X=2)= \frac{3!}{2!(3-2)!}(0.5)^2(1-0.5)^{3-2}$ => $P(X=2)= 3(0.5)^2(0.5)^{1}$ =>  $P(X=2)= 3(0.5)^3$ =>  $P(X=2)= 0.375$

Entonces tenemos que la probabilidad de tener dos hijos varones es de 37.5%

**b) ¿Cuál es el número esperado (media) de hijos varones?**
    
Utilizando $\mu=np$ tenemos que $\mu$ es igual a 1.5.   

**c) ¿Cuál es la varianza del número de hijos varones?**
    
Utilizando $\sigma^2=np(1-p)$ y sustituyendo los valores, tenemos qeu la varianza es de 0.75

### 6. Correlación de variables utilizando Python
### Utiliza el conjunto de datos load_breast_cancer de Scikit-learn para identificar qué característica tiene la correlación lineal más fuerte (positiva o negativa) con la variable objetivo target.


```python

import numpy as np
import pandas as pd
from sklearn.datasets import load_breast_cancer

# Cargar los datos
data = load_breast_cancer()

# Crear un DataFrame
df = pd.DataFrame(data.data, columns=data.feature_names)

# Agregar la variable objetivo
df["target"] = 1 - data.target

# a)
correlaciones = df.corr(method="pearson")["target"]

# Mostrar las correlaciones de todas las características
print("a) Correlación de Pearson con target:")
print(correlaciones.drop("target").sort_values(ascending=False))

# b) 
correlaciones_sin_target = correlaciones.drop("target")
caracteristica = correlaciones_sin_target.abs().idxmax()
valor = correlaciones_sin_target[caracteristica]

print("\nb) Característica con mayor correlación absoluta:")
print("Característica:", caracteristica)
print("Correlación:", valor)

# c)
print("\nc) Interpretación:")
print("La correlación es positiva.")
print("Al aumentar la característica, target tiende a aumentar.")
```

### Conclusión:
Mediante el coeficiente de correlación, se pudo identificar la caracteristica con mayor valor lo que permite reconocer cuál presenta la asociación más fuerte con la clasificación del tumor. Por ultimo, este análisis permite identificar características relevantes para comprender los datos, aunque la correlación no implica una clasificación correcta.
