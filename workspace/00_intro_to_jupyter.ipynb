## 1. ¿Qué es un Jupyter Notebook?

Un **notebook** es un documento interactivo que mezcla en un solo archivo:

- 🧑‍💻 **Código** que puedes ejecutar por partes.
- 📊 **Resultados** de ese código: texto, tablas, gráficas…
- 📝 **Texto explicativo** con formato, imágenes y ecuaciones.

Por eso es la herramienta estándar en ciencia de datos: permite **explorar, documentar y compartir** un análisis en el mismo lugar.

El nombre *Jupyter* viene de **Ju**lia, **Py**thon y **R**, los tres primeros lenguajes que soportó. Los archivos tienen extensión **`.ipynb`** (*IPython Notebook*) y por dentro son un archivo JSON.

## 2. Celdas: los bloques de un notebook

Un notebook está formado por **celdas**. Hay dos tipos principales:

- **Código**: contiene Python. Al ejecutarla, el resultado aparece justo debajo.
- **Markdown**: contiene texto con formato, como esta celda.

### Ejecutar una celda

Selecciona la siguiente celda y pulsa **`Shift + Enter`**:

````python
print("¡Hola, mundo! 👋")
````

Hay tres formas de ejecutar una celda:

| Atajo | Qué hace |
|---|---|
| `Shift + Enter` | Ejecuta la celda y pasa a la siguiente |
| `Ctrl + Enter` (`Cmd + Enter` en macOS) | Ejecuta la celda y se queda en ella |
| `Alt + Enter` (`Option + Enter` en macOS) | Ejecuta la celda y crea una nueva debajo |

### La última línea se muestra sola

Si la **última línea** de una celda es una expresión, Jupyter muestra su valor automáticamente, sin necesidad de `print`:

````python
a = 3
b = 4
a + b
````

Pero **solo la última**. Aquí únicamente verás el resultado de `a * b`:

````python
a - b
a * b
````

Si quieres ver varios valores, usa `print()`.

> 💡 **Truco:** termina la última línea con `;` para ocultar su salida. Te será útil con las gráficas.

````python
print(a - b)
print(a * b)
````

## 3. Modo comando y modo edición

Jupyter tiene dos modos, igual que el editor *vim*:

- ✏️ **Modo edición** (`Enter` o clic dentro de la celda): escribes dentro de la celda.
- ⌨️ **Modo comando** (`Esc`): trabajas con las celdas como bloques (crear, borrar, mover…).

### Atajos en modo comando

| Tecla | Acción |
|---|---|
| `A` | Insertar celda **arriba** (*above*) |
| `B` | Insertar celda **abajo** (*below*) |
| `D`, `D` | Borrar la celda (pulsa `D` dos veces) |
| `Z` | Deshacer el borrado de una celda |
| `M` | Convertir la celda a **Markdown** |
| `Y` | Convertir la celda a **código** |
| `↑` / `↓` | Moverte entre celdas |
| `Shift + ↑/↓` | Seleccionar varias celdas |

> **En Google Colab** los atajos de modo comando van precedidos de `Ctrl + M` (`Cmd + M` en macOS). Por ejemplo: `Ctrl + M`, `B` para insertar una celda abajo.

✍️ **Practica:** pulsa `Esc`, luego `B` para crear una celda nueva debajo de esta, escribe `print("mi primera celda")` y ejecútala.

## 4. El kernel y el estado del notebook

El **kernel** es el proceso de Python que ejecuta tu código en segundo plano. Todas las celdas comparten el mismo kernel, así que **las variables que defines en una celda existen en todas las demás**:

````python
nombre = "Ada Lovelace"
````

````python
f"La primera programadora de la historia fue {nombre}"
````

### ⚠️ El orden de ejecución importa

El número a la izquierda de cada celda, `[1]`, `[2]`, …, indica **en qué orden se ejecutó**. El kernel recuerda el **orden en que ejecutaste** las celdas, **no** el orden en que aparecen en la página.

Haz este experimento:

1. Ejecuta las dos celdas siguientes en orden. Verás `20`.
2. Cambia `x = 10` por `x = 100`, pero **no** ejecutes esa celda.
3. Vuelve a ejecutar solo la segunda celda. Sigue mostrando `20`: el kernel no sabe que cambiaste el texto.

````python
x = 10
````

````python
x * 2
````

Esta es la fuente de error **#1** al trabajar con notebooks: el código que ves no coincide con el estado del kernel.

### Reiniciar el kernel

Cuando algo se comporte raro, reinicia y ejecuta todo desde cero:

| Herramienta | Menú |
|---|---|
| Jupyter Lab | **Kernel → Restart Kernel and Run All Cells…** |
| VS Code | Botón **Restart**, luego **Run All** |
| Google Colab | **Entorno de ejecución → Reiniciar y ejecutar todo** |

> ✅ **Regla de oro:** antes de entregar o compartir un notebook, haz *Restart & Run All* y comprueba que funciona de principio a fin.

### Ecuaciones con LaTeX

En Machine Learning escribiremos muchas fórmulas. Jupyter las muestra con **LaTeX**:

- **En línea**, entre `$...$`: el modelo lineal es $\hat{y} = w x + b$.
- **En bloque**, entre `$$...$$`:

$$
\text{MSE} = \frac{1}{n} \sum_{i=1}^{n} \left( y_i - \hat{y}_i \right)^2
$$

✍️ **Practica:** crea una celda Markdown (`Esc`, `B`, `M`) con tu nombre como título y la fórmula de la media: $\bar{x} = \frac{1}{n}\sum_{i=1}^{n} x_i$.

## 7. Comandos mágicos

Los **comandos mágicos** son instrucciones especiales de Jupyter (no de Python). Empiezan por:

- `%`: afectan a **una línea**.
- `%%`: afectan a **toda la celda**. Deben ir en la primera línea.

### Medir tiempos

Muy útiles para comparar qué código es más rápido:

````python
%timeit sum(range(1_000))
````

````python
%%time
total = 0
for i in range(1_000_000):
    total += i
total
````

### Inspeccionar variables

`%who` lista las variables definidas y `%whos` añade su tipo y su valor:

````python
%whos
````

### Navegar por archivos

````python
%pwd
````

Para ver la lista completa de comandos mágicos: `%lsmagic`.

````python
%lsmagic
````

## 9. Salida enriquecida

Los notebooks pueden mostrar mucho más que texto plano. La función `display` de IPython permite mostrar varios objetos en una misma celda, con formato:

````python
from IPython.display import HTML, Markdown, display

display(Markdown("### Esto es Markdown generado desde Python"))
display(Markdown(f"El valor de `x` es **{x}**"))
display(HTML("<p style='color: tomato;'>Y esto es HTML 🎨</p>"))
````

Más adelante usaremos esta capacidad para mostrar tablas de datos con **pandas** y gráficas con **matplotlib**, directamente en el notebook.

## 10. Errores: cómo leer un *traceback*

Cuando algo falla, Python muestra un **traceback**. Asusta al principio, pero tiene una estructura clara.

**Léelo de abajo hacia arriba:**

1. La **última línea** te dice el **tipo de error** y el mensaje.
2. Las líneas anteriores indican **dónde** ocurrió (la flecha `---->` señala la línea exacta).

Ejecuta la siguiente celda:

````python
def dividir(a, b):
    return a / b

dividir(10, 0)
````

La última línea dice `ZeroDivisionError: division by zero`: intentamos dividir entre cero.

Un error **no rompe el notebook**: el kernel sigue funcionando y las variables siguen ahí. Corrige la celda y vuelve a ejecutarla.

> 💡 Si una celda se queda ejecutándose para siempre (`[*]`), interrúmpela con el botón ⏹ o, en modo comando, pulsando `I`, `I`.

## 11. Buenas prácticas

1. **Imports al principio.** Pon todos los `import` en las primeras celdas.
2. **Restart & Run All** antes de compartir. Si no funciona de arriba abajo, no es reproducible.
3. **Celdas cortas** que hagan una sola cosa.
4. **Documenta con Markdown** el *porqué*, no solo el *qué*.
5. **No guardes contraseñas ni API keys** en un notebook. Quedan en el archivo y en el historial de Git.
6. **Limpia las salidas antes de hacer commit** (*Edit → Clear Outputs of All Cells*). Así los diffs son más legibles y el repo pesa menos.
7. **Notebooks para explorar, scripts `.py` para reutilizar.** Cuando una función se use en varios notebooks, muévela a un módulo de Python.

### Varianza
**La varianza mide que tanto estan dispersos tus datos o su fiabilidad, denominados con la letra $\sigma$ y utilizando la siguiente ecuacion podemos conocer esa dispercion**

$\sigma^{2} = \sum^{n}_{i=1} (x_{i}-\bar{x})^2$
